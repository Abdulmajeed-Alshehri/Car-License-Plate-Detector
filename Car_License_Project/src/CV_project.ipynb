```python
import cv2
from ultralytics import YOLO
```

```python
#Copy Your Video Path Here
vid_path = r""

cap = cv2.VideoCapture(vid_path)

#Copy best.pt Path Here
model = YOLO(r"")

#Adjust The Width Here
width = 800

#Adjust The Height Here
height = 600
```

```python
if not cap.isOpened():
    print("ERROR")
    exit()

while cap.isOpened():
    success , frame = cap.read()
    if success:
        resized_frame = cv2.resize(frame,(width,height))

        #You Can Lower Or Higher The Conf In This Line
        result = model.predict(source=resized_frame,conf = 0.4,show_labels=False)
        annotated_frame = result[0].plot(labels=False)

        cv2.imshow("Video",annotated_frame)
        if cv2.waitKey(1) & 0xFF == ord("q"):
            break

    else:
        break

cap.release()
cv2.destroyAllWindows()
```
